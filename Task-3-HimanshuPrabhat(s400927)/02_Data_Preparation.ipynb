# FIFA World Cup 2026 — Objective 1, Task 3
## Raw Population → Final Sample

### Purpose
This notebook only shows how the final 36-team sample is created from the 48-team raw population.

It does **not** perform the main Task 3 statistical analysis.

Process:

1. Load the raw 48-team population
2. Calculate Shooting Accuracy
3. Split teams into Advanced and Eliminated groups
4. Apply 75% proportionate stratified random sampling
5. Show the 36 selected teams
6. Show the 12 teams not selected

## 1. Load the raw population

```python
import pandas as pd

raw = pd.read_csv("Task3_Raw_Population.csv")

print("Raw population rows:", len(raw))
print("\nOutcome counts:")
print(raw["Outcome"].value_counts())

raw.head()
```

## 2. Calculate Shooting Accuracy

The website data gives:
- Total Shots
- Shots on Target

Python calculates:

**Shooting Accuracy (%) = (Shots on Target / Total Shots) × 100**

```python
population = raw.copy()

population["Shooting_Accuracy"] = (
    population["Shots_on_Target"] / population["Total_Shots"] * 100
)

population[
    ["Group", "Team", "Outcome", "Total_Shots",
     "Shots_on_Target", "Shooting_Accuracy"]
].head(10)
```

## 3. Split the population into Advanced and Eliminated teams

Population:
- 32 Advanced teams
- 16 Eliminated teams

```python
advanced_population = population[
    population["Outcome"] == "Advanced"
]

eliminated_population = population[
    population["Outcome"] == "Eliminated"
]

print("Advanced population:", len(advanced_population))
print("Eliminated population:", len(eliminated_population))
```

## 4. Create the final 36-team sample

We select 75% from each group:

- 24 of 32 Advanced teams
- 12 of 16 Eliminated teams

`random_state=42` is fixed so the same teams are selected every time the notebook is run.

```python
advanced_sample = advanced_population.sample(
    n=24,
    random_state=42
)

eliminated_sample = eliminated_population.sample(
    n=12,
    random_state=42
)

sample = pd.concat(
    [advanced_sample, eliminated_sample],
    ignore_index=True
)

print("Final sample size:", len(sample))
print("\nSample outcome counts:")
print(sample["Outcome"].value_counts())
```

## 5. Final 36 selected teams

```python
sample[
    ["Group", "Team", "Outcome", "Total_Shots",
     "Shots_on_Target", "Shooting_Accuracy"]
].sort_values(["Outcome", "Group", "Team"])
```

## 6. 12 teams not selected

These teams are **not bad data**.

They are simply the 12 teams that were not selected by the 75% proportionate stratified random sampling process.

```python
excluded = population[
    ~population["Team"].isin(sample["Team"])
].copy()

excluded["Reason"] = (
    "Not selected by the 75% proportionate stratified random sample "
    "(random_state=42)."
)

print("Teams not selected:", len(excluded))

excluded[
    ["Group", "Team", "Outcome", "Shooting_Accuracy", "Reason"]
].sort_values(["Outcome", "Group", "Team"])
```

## Final result

The bridge step is complete:

**48-team raw population → Shooting Accuracy calculated → 36-team final sample**

The main Task 3 notebook can now use this 36-team sample for the actual statistical analysis.
